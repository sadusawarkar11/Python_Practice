# Loops — Complete Recap

### 1. `for` Loop
Iterates over any **iterable** such as lists, strings, dictionaries, sets, tuples, `range()`, `zip()`, and `enumerate()`.

```python
for x in [1, 2, 3]:
    print(x)
```

### 2. `while` Loop
Runs repeatedly while a condition is `True`.

```python
while n > 0:
    n -= 1
```

It can run indefinitely if the condition never becomes `False`, unless a `break` is used.

### 3. `range(start, stop, step)`
Generates a sequence of numbers.

```python
range(1, 5)       # 1, 2, 3, 4
range(5, 0, -1)   # 5, 4, 3, 2, 1
```

**`stop` is always exclusive.**

### 4. `break`
Immediately exits the **entire current loop**.

```python
for i in range(10):
    if i == 5:
        break
```

### 5. `continue`
Skips the current iteration and moves to the next one.

```python
for i in range(5):
    if i == 2:
        continue
    print(i)
```

### 6. Loop `else`
The `else` block runs only when the loop finishes **without encountering `break`**.

```python
for i in range(5):
    print(i)
else:
    print("Completed")
```

### 7. Nested Loops
`break` and `continue` affect only the **innermost loop** in which they appear.

### 8. `enumerate()`
Provides both the **index and value** while iterating.

```python
for index, value in enumerate(["a", "b", "c"]):
    print(index, value)
```

### 9. `zip()`
Iterates over multiple iterables **in parallel** and stops at the shortest one.

```python
for a, b in zip([1, 2], ["a", "b"]):
    print(a, b)
```

### 10. Iterating Over Dictionaries

By default, a dictionary loop iterates over its **keys**:

```python
for key in d:
    print(key)
```

For explicit control:

```python
d.keys()     # keys
d.values()   # values
d.items()    # key-value pairs
```

### 11. Mutating a List While Iterating
Changing a list while looping over it can cause elements to be **skipped or processed unexpectedly**.

```python
for x in my_list:
    my_list.remove(x)   # ⚠️ risky
```

### 12. Loop Variable Scope
Python's loop variable remains available after the loop:

```python
for i in range(3):
    pass

print(i)   # 2
```

### 13. Walrus Operator in `while`

`:=` can assign and test a value in the same condition:

```python
while (line := f.readline()):
    print(line)
```

### 14. List/Dictionary Comprehensions
Provide a concise, Pythonic way to build collections using loop-like logic.

```python
squares = [x * x for x in range(5)]

d = {x: x * x for x in range(5)}
```

### 15. Iterating Over Strings
A string is iterable, so a `for` loop processes it **one character at a time**.

```python
for char in "Python":
    print(char)
```

###  Quick Summary

# `for-else` with `break`

```python
for i in range(5):
    if i == 1:
        break
    print(i)
else:
    print("Done")
```

### Trace

```text
i = 0 → prints 0
i = 1 → break → loop exits immediately
```

Because the loop exited using `break`, the `else` block **does not run**.

### Output

```text
0
```

###  Key Rule

```text
Loop finishes normally → else runs
Loop exits with break   → else does NOT run
```

> **The `else` belongs to the loop, not to the `if`.**
> It runs only when the loop completes without encountering `break`.

````python
words = ["cat", "dog", "elephant", "fox"]
for i, word in enumerate(words):
    if len(word) > 5:
        continue
    print(i, word)

# 0 cat
# 1 dog
# 3 fox
````

````python
names = ["Alice", "Bob"]
ages = [25, 30, 35]

for name, age in zip(names, ages):
    print(name, age)

# Alice 25
# Bob 30
````

````python
matrix = [[1, 2], [3, 4], [5, 6]]
for row in matrix:
     for val in row:
         if val == 4:
             break
         print(val)

# 1
# 2
# 3
# 5
# 6
````

````python
nums = [1, 2, 3, 4, 5]
for n in nums:
    if n == 3:
        nums.remove(n)
    print(n)

# 1
# 2
# 3
# 5
````

````python
squares = [x**2 for x in range(5) if x % 2 == 0]
print(squares)

# [0, 4, 16]
````

````python
d = {"a": 1, "b": 2, "c": 3}
for key in d:
    d[key] *= 10
print(d)

# {"a": 10, "b": 20, "c": 30}
````

````python
funcs = []
for i in range(3):
    funcs.append(lambda: i)

results = [f() for f in funcs]
print(results)

# [0,1,2]
````

# Late-Binding Closure Gotcha

```python
funcs = [lambda: i for i in range(3)]

print([f() for f in funcs])
```

### Output

```text
[2, 2, 2]
```

### Why?

The lambda:

```python
lambda: i
```

does **not capture the current value** of `i`.

Instead, it captures the **variable `i`** and looks up its value when the lambda is called.

The loop finishes first:

```text
i = 0
i = 1
i = 2
```

After the loop, `i` is `2`.

Then the lambdas are called, and all of them look up the same `i`:

```text
lambda → i → 2
lambda → i → 2
lambda → i → 2
```

Therefore:

```text
[2, 2, 2]
```

###  Key Takeaway

> **Closures capture variables, not their values at creation time.**

If you want each lambda to remember its current value, use a default argument:

```python
funcs = [lambda i=i: i for i in range(3)]

print([f() for f in funcs])
```

Output:

```text
[0, 1, 2]
```

# Loop Variable Scope — Overwriting an Existing Variable

```python
i = 10

for i in range(3):
    print(i)

print(i)
```

### Output

```text
0
1
2
2
```

### Why?

Python's `for` loop does **not create a new scope**.

The loop reuses the existing variable `i`:

```text
Before loop → i = 10

i = 0 → prints 0
i = 1 → prints 1
i = 2 → prints 2

After loop → i = 2
```

So the loop **overwrites** the previous value `10`.

###  Key Takeaway

> **Python loops do not create their own scope. The loop variable remains available after the loop and keeps its final value.**

```python
i = 10

for i in range(3):
    pass

print(i)   # 2
```

````python
count = 0
while count < 5:
    count += 1
    if count == 3:
        continue
    print(count)
    if count == 4:
        break

# 1
# 2
# 4
````

````python
n = 0
while n < 3:
    n += 1
else:
    print("Loop finished, n =", n)
print("After loop, n =", n)

# Loop finished, n =3
# After loop, n =3
````
