### Bitwise NOT (`~`) Operator

The `~` operator (bitwise NOT) flips all bits of a number.

### Easy Rule to Remember

```python
~x = -(x + 1)
```

So,

```python
~5
# -(5 + 1)
# -6
```

### Why Isn't `~5` Equal to `2`?

Python integers do **not** have a fixed number of bits. They use **infinite-precision signed integers**, so the easiest and correct way is to remember the formula instead of trying to flip the visible binary digits.

### Examples

```python
~5     # -6
~0     # -1
~10    # -11
~(-1)  # 0
```

### Key Point

Always use this formula:

```python
~x = -(x + 1)
```

**One-line summary:**
> The `~` operator returns `-(x + 1)`, so `~5` is `-6`, not `2`.

### `and` and `or` Return Values (Not Just `True` or `False`)

Python's `and` and `or` operators return **actual values**, not always `True` or `False`.

### `and` Rule
- Checks values from left to right.
- Returns the **first falsy value** it finds.
- If all values are truthy, it returns the **last value**.

Examples:

```python
5 and 3      # 3
0 and 3      # 0
```

### `or` Rule
- Checks values from left to right.
- Returns the **first truthy value** it finds.
- If all values are falsy, it returns the **last value**.

Examples:

```python
5 or 3       # 5
0 or 3       # 3
```

### Short-Circuit Evaluation
Python stops checking as soon as it knows the result.

Example:

```python
name = user_input or "Guest"
```

If `user_input` is empty or falsy, `name` becomes `"Guest"`.

### Key Point
- `and` → First **falsy** value, otherwise **last** value.
- `or` → First **truthy** value, otherwise **last** value.

If you want only `True` or `False`, use `bool()`:

```python
bool(5 and 3)   # True
```

**One-line summary:**
> `and` and `or` return actual values, not just booleans. `and` returns the first falsy value, while `or` returns the first truthy value.

### Chained Comparison with `True`

Python treats:

```python
x == 10 == True
as:

(x == 10) and (10 == True)

Example (x = 10):

x == 10 → True
10 == True → 10 == 1 → False (because True is equal to 1)
Final result → True and False → False
Key Point

True is not a special confirmation. It is just another value (1) in a chained comparison.

Similarly,

x == 10 == 1

becomes:

(x == 10) and (10 == 1)

→ True and False → False

One-line summary:
> In a chained comparison, `True` is treated like the value `1`, not as a keyword that confirms the previous comparison.
