#  **Polymorphism**

### 1. Polymorphism

**Same method call → different behavior** depending on the object's type.

```python
class Bird:
    def fly(self):
        print("Bird is flying")


class Penguin:
    def fly(self):
        print("Penguin cannot fly")


for animal in [Bird(), Penguin()]:
    animal.fly()
```

The same:

```python
animal.fly()
```

behaves differently for each object.

---

### 2. Duck Typing

Python often cares about **what an object can do**, not what type it is.

```python
class Dog:
    def speak(self):
        print("Woof")


class Cat:
    def speak(self):
        print("Meow")


def make_speak(animal):
    animal.speak()
```

No inheritance is required.

```python
make_speak(Dog())
make_speak(Cat())
```

If the object has the required `speak()` method, Python can use it.

> **"If it walks like a duck and quacks like a duck, treat it like a duck."**

---

### 3. Built-in Polymorphism

Built-in functions/operators can behave differently depending on the object.

```python
len([1, 2, 3])       # 3
len("Hello")         # 5

[1, 2] + [3, 4]      # [1, 2, 3, 4]
"Hello" + " World"   # "Hello World"
```

This works because Python objects define how these operations should behave using **dunder methods**.

---

### 4. Functions Can Accept Different Types

A function can simply expect an object to provide the required method.

```python
def start(vehicle):
    vehicle.start()
```

It doesn't need to check:

```python
isinstance(vehicle, Car)
```

As long as `vehicle.start()` exists, it works.

---

### 5. `hasattr()`

`hasattr()` checks whether an object has a particular attribute or method.

```python
if hasattr(obj, "speak"):
    obj.speak()
```

It returns:

```text
True  → attribute exists
False → attribute doesn't exist
```

---

### 6. Duck Typing vs Common Base Class

**Polymorphism through inheritance:**

```text
Animal
 ├── Dog
 └── Cat
```

Both share a common parent.

**Pure duck typing:**

```text
Dog    → has speak()
Robot  → has speak()
Person → has speak()
```

They don't need any inheritance relationship.

Python commonly favors this flexible **duck-typing** approach.

---

### 7. `AttributeError` — Duck Typing's Tradeoff

If the required method doesn't exist:

```python
def make_speak(animal):
    animal.speak()

make_speak(5)
```

Python raises:

```text
AttributeError
```

because `int` doesn't have a `speak()` method.

###  Key Takeaway

> **Polymorphism = same interface, different behavior.**
>
> **Duck typing = Python cares about what an object can do, not what type it belongs to.**

```text
Inheritance → "What type are you?"
Duck typing  → "What can you do?"
```

````python
class Cat:
    def speak(self):
        return "Meow"

class Dog:
    def speak(self):
        return "Woof"

animals = [Cat(), Dog()]
for a in animals:
    print(a.speak())

# Meow
# Woof
````

````python
def describe(obj):
    return obj.name

class Product:
    name = "Widget"

class Employee:
    name = "Alice"

print(describe(Product()))
print(describe(Employee()))

````
