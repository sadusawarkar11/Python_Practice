# Classes & Objects Basics — Recap

### 1. `class`
A **class** is a blueprint used to create objects.

```python
class Student:
    pass
```

### 2. `__init__`
`__init__` runs automatically when an object is created and is used to set up its initial state.

```python
class Student:
    def __init__(self, name):
        self.name = name
```

### 3. `self`
`self` refers to the **specific object (instance)** calling the method.

```python
class Student:
    def greet(self):
        print("Hello")
```

`self` must be the first parameter of an instance method by convention.

### 4. Instance Attributes
Created using `self.attribute`.

```python
class Student:
    def __init__(self, name):
        self.name = name
```

Each object gets its **own separate** `name`.

```python
s1 = Student("Alice")
s2 = Student("Bob")
```

```text
s1.name → Alice
s2.name → Bob
```

### 5. Class Attributes
Defined directly inside the class and shared by all instances.

```python
class Student:
    school = "ABC School"
```

```python
s1 = Student()
s2 = Student()

print(s1.school)   # ABC School
print(s2.school)   # ABC School
```

### 6. Instantiation
Creating an object from a class is called **instantiation**.

```python
student = Student("Alice")
```

Python automatically calls:

```python
Student.__init__(student, "Alice")
```

### 7. Attribute Lookup Order
When you access:

```python
obj.attribute
```

Python first checks the **instance's attributes**.

If it doesn't find the attribute there, it looks in the **class**.

```text
Instance __dict__
       ↓
Class attributes
```

### 8. Mutable Class Attribute Gotcha

A mutable class attribute such as a list or dictionary is **shared by all instances**.

```python
class Student:
    subjects = []

s1 = Student()
s2 = Student()

s1.subjects.append("Python")

print(s2.subjects)
```

Output:

```text
['Python']
```

Why?

Both objects refer to the **same class-level list**.

If every object needs its own list, create it as an instance attribute:

```python
class Student:
    def __init__(self):
        self.subjects = []
```

Now:

```text
s1.subjects → separate list
s2.subjects → separate list
```

###  Key Takeaway

| Concept | Meaning |
|---|---|
| `class` | Blueprint for objects |
| `__init__` | Initializes an object |
| `self` | Refers to the current instance |
| Instance attribute | Unique to each object |
| Class attribute | Shared through the class |
| Instantiation | Creating an object |
| Attribute lookup | Instance → Class |
| Mutable class attribute | ⚠️ Shared by instances |

> **The most important distinction:**  
> `self.attr` → usually **instance-specific**  
> `attr` directly inside the class → **class-level/shared**

````python
class Dog:
    def __init__(self, name):
        self.name = name

d = Dog("Rex")
print(d.name)

# Rex
````

````python
class Counter:
    count = 0

c1 = Counter()
c2 = Counter()

print(c1.count, c2.count)

# 0 0
````

````python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def introduce(self):
        return f"I am {self.name}, {self.age} years old"

p = Person("Sam", 25)
print(p.introduce())

# I am Sam, 25 years old
````

````python
class Box:
    def __init__(self):
        self.items = []

b1 = Box()
b2 = Box()

b1.items.append("book")
print(b1.items, b2.items)

# ['book']  []
````

````python
class Robot:
    def __init__(self, name):
        self.name = name

    def rename(self, new_name):
        self.name = new_name
        return self

r = Robot("R1")
r.rename("R2").rename("R3")
print(r.name)

# R3
````

````python
class Team:
    members = []

    def __init__(self, name):
        self.name = name
        Team.members.append(name)

t1 = Team("Alice")
t2 = Team("Bob")

print(t1.members)
print(t2.members)
print(Team.members)

# ['Alice']
# ['Alice', 'Bob']
# ['Alice', 'Bob']
````

# Mutable Class Attribute — Shared List

```python
class Team:
    members = []

    def __init__(self, name):
        Team.members.append(name)


t1 = Team("Alice")
t2 = Team("Bob")

print(t1.members)
print(t2.members)
print(Team.members)
```

### Output

```text
['Alice', 'Bob']
['Alice', 'Bob']
['Alice', 'Bob']
```

### Why?

First:

```python
t1 = Team("Alice")
```

adds `"Alice"`:

```text
['Alice']
```

Then:

```python
t2 = Team("Bob")
```

adds `"Bob"` to the **same list**:

```text
['Alice', 'Bob']
```

Because `members` is a **class attribute**, there is only **one shared list**.

Therefore:

```text
t1.members   ─┐
t2.members   ─┼→ same list
Team.members  ─┘
```

By the time the `print()` statements execute, both names have already been added.

###  Key Takeaway

> **A mutable class attribute is shared by all instances unless each instance creates its own copy.**

This is similar to the **mutable default argument** problem in functions: one mutable object is unintentionally shared across multiple uses.

````python
class Config:
    settings = {"debug": False}
c1 = Config()
c2 = Config()
c1.settings["debug"] = True
print(c1.settings, c2.settings)
print(c1.settings is c2.settings)

# {"debug": True} {"debug": True}
# True
````

````python
class Node:
    def __init__(self, value, next_node=None):
        self.value = value
        self.next_node = next_node

n3 = Node(3)
n2 = Node(2, n3)
n1 = Node(1, n2)

current = n1
while current:
    print(current.value)
    current = current.next_node

# 1
# 2
# 3
````

````python
class BankAccount:
    def __init__(self, balance=0):
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

accounts = [BankAccount(100), BankAccount(200)]

for acc in accounts:
    acc.deposit(50)

total = sum(acc.balance for acc in accounts)
print(total)

# 400
````

````python
class Warehouse:
    def __init__(self):
        self.stock = []

def add_stock(warehouse, item):
    warehouse.stock += [item]

w = Warehouse()
add_stock(w, "apple")
print(w.stock)
print(id(w.stock))

def add_stock_v2(warehouse, item):
    warehouse.stock = warehouse.stock + [item]

add_stock_v2(w, "banana")
print(w.stock)

# ['apple']
# some id number
# ['apple','banana']
````

````python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

p1 = Point(1, 2)
p2 = Point(1, 2)

print(p1 == p2)
print(p1 is p2)
print(p1.__dict__ == p2.__dict__)

# False
# False
# True
````

````python
class Animal:
    sounds = []

class Dog(Animal):
    pass

class Cat(Animal):
    pass

Dog.sounds.append("Woof")
Cat.sounds.append("Meow")

print(Dog.sounds)
print(Cat.sounds)
print(Animal.sounds)

# ['Woof','Meow']
# ['Woof','Meow']
# ['Woof','Meow']
````

````python
class Wallet:
    def __init__(self, balance):
        self.balance = balance

def give_bonus(wallet):
    w = Wallet(wallet.balance + 100)

w = Wallet(50)
give_bonus(w)
print(w.balance)

# 150
````

# Reassignment vs. Mutation in Functions

```python
def give_bonus(wallet):
    wallet = Wallet(wallet.balance + 100)

w = Wallet(500)

give_bonus(w)

print(w.balance)
```

### Output

```text
500
```

### Why?

Inside the function:

```python
wallet = Wallet(wallet.balance + 100)
```

creates a **new `Wallet` object** and reassigns the local variable `wallet` to it.

It does **not** modify the original object `w`.

```text
Before:
w ─────────→ Wallet(500)

Inside function:
wallet ────→ Wallet(500)

wallet = Wallet(600)

wallet ────→ NEW Wallet(600)
w ─────────→ ORIGINAL Wallet(500)
```

The new `Wallet(600)` is never returned, so it is discarded when the function finishes.

### Mutation vs Reassignment

**Mutation:**

```python
wallet.balance += 100
```

→ modifies the existing object → `w.balance` becomes `600`.

**Reassignment:**

```python
wallet = Wallet(wallet.balance + 100)
```

→ creates a new object and changes only the local variable → `w.balance` remains `500`.

###  Key Takeaway

> **Mutating the object affects the caller's object. Reassigning the local parameter does not.**
