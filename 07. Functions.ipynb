# **Functions**

### 1. Basic Function Definition

Use `def` to create a function.

```python
def add(a, b):
    return a + b
```

A function can return a value using `return`.

If there is no `return`, Python implicitly returns `None`.

```python
def greet():
    print("Hello")

print(greet())   # None
```

---

### 2. Positional vs Keyword Arguments

**Positional arguments** are matched by position:

```python
def add(a, b):
    return a + b

add(1, 2)
```

**Keyword arguments** specify parameter names:

```python
add(a=1, b=2)
```

---

### 3. Default Arguments

A parameter can have a default value:

```python
def greet(name, message="Hello"):
    print(message, name)

greet("Sam")
greet("Sam", "Hi")
```

Avoid mutable default arguments such as:

```python
def add_item(item, items=[]):   # wrong
    items.append(item)
    return items
```

The same list is reused between calls.

Prefer:

```python
def add_item(item, items=None):
    if items is None:
        items = []
    items.append(item)
    return items
```

---

### 4. `*args`

Collects extra **positional arguments** into a tuple.

```python
def add(*args):
    print(args)

add(1, 2, 3)
# (1, 2, 3)
```

---

### 5. `**kwargs`

Collects extra **keyword arguments** into a dictionary.

```python
def show(**kwargs):
    print(kwargs)

show(name="Sam", age=20)
# {'name': 'Sam', 'age': 20}
```

---

### 6. Argument Order

The general order is:

```text
positional → *args → keyword/default → **kwargs
```

For example:

```python
def f(a, b=10, *args, **kwargs):
    pass
```

---

### 7. Keyword-Only Arguments

Parameters after `*` must be supplied using keywords.

```python
def greet(name, *, age):
    print(name, age)

greet("Sam", age=20)   # right
greet("Sam", 20)       # wrong
```

---

### 8. Positional-Only Arguments

Parameters before `/` can only be passed positionally.

```python
def add(a, b, /):
    return a + b

add(1, 2)        # right
add(a=1, b=2)    # wrong
```

Available since **Python 3.8**.

---

### 9. `lambda`

A `lambda` is a small anonymous function containing one expression.

```python
square = lambda x: x * x

print(square(5))   # 25
```

---

### 10. Variable Scope — LEGB

Python searches for variables using the **LEGB** rule:

```text
L → Local
E → Enclosing
G → Global
B → Built-in
```

`global` allows modifying a global variable:

```python
x = 10

def change():
    global x
    x = 20
```

`nonlocal` allows modifying a variable from an enclosing function:

```python
def outer():
    x = 10

    def inner():
        nonlocal x
        x = 20
```

---

### 11. Closures

A closure is an inner function that **remembers variables from its enclosing scope**.

```python
def outer(x):
    def inner():
        return x
    return inner

f = outer(10)

print(f())   # 10
```

---

### 12. First-Class Functions

Functions are objects in Python, so they can be:

- Stored in variables
- Passed as arguments
- Returned from functions
- Stored in lists/dictionaries

```python
def square(x):
    return x * x

func = square

print(func(5))   # 25
```

---

### 13. Recursion

A recursive function **calls itself**.

It needs:

- A **base case** → stops recursion
- A **recursive case** → calls itself again

```python
def factorial(n):
    if n == 0:          # base case
        return 1
    return n * factorial(n - 1)
```

---

### 14. Docstrings & Type Hints

A **docstring** describes what a function does:

```python
def add(a, b):
    """Return the sum of two numbers."""
    return a + b
```

Type hints describe expected types:

```python
def add(a: int, b: int) -> int:
    return a + b
```

Type hints are mainly for readability, tooling, and static analysis; Python does not enforce them automatically.

---

### 15. Argument Unpacking

Use `*` to unpack a sequence into positional arguments:

```python
def add(a, b, c):
    return a + b + c

nums = [1, 2, 3]

print(add(*nums))   # 6
```

Use `**` to unpack a dictionary into keyword arguments:

```python
def greet(name, age):
    print(name, age)

data = {"name": "Sam", "age": 20}

greet(**data)
```

# Function Without `return`

```python
def greet(name):
    print(f"Hello, {name}")

result = greet("Alice")
print(result)
```

### Output

```text
Hello, Alice
None
```

### Why?

Inside the function:

```python
print(f"Hello, {name}")
```

prints:

```text
Hello, Alice
```

But the function has **no `return` statement**.

Therefore, Python automatically returns:

```python
None
```

So:

```python
result = greet("Alice")
```

means:

```text
result = None
```

Then:

```python
print(result)
```

prints:

```text
None
```

###  Key Takeaway

> A function can **print something** without **returning something**.

```text
print()  → displays a value
return   → sends a value back to the caller
```

If a function has no `return`, its return value is automatically **`None`**.

````python
def add(a, b=10):
    return a + b

print(add(5))
print(add(5, 20))

# 15
# 25
````

````python
def multiply(a, b):
    return a * b

result = multiply(b=3, a=4)
print(result)

# 12
````

````python
square = lambda x: x ** 2
print(square(5))

# 25
````

````python
def show(*args, **kwargs):
    print(args)
    print(kwargs)

show(1, 2, 3, name="Alice", age=25)

# (1, 2, 3)
# {'name': 'Alice', 'age': 25}
````

````python
def f(a, b, *, c):
    return a + b + c

print(f(1, 2, c=3))
print(f(1, 2, 3))

# 6
````

````python
def outer():
    x = 10
    def inner():
        print(x)
    inner()

outer()

# 10
````

````python
def apply(func, value):
    return func(value)

print(apply(lambda x: x * 2, 5))
print(apply(str, 100))

# 10
# '100'
````

````python
def add_item(item, lst=[]):
    lst.append(item)
    return lst

print(add_item(1))
print(add_item(2))
print(add_item(3, []))
print(add_item(4))

# [1]
# [1,2]
# [3]
# [1,2,4]
````

````python
def make_counter():
    count = 0
    def counter():
        nonlocal count
        count += 1
        return count
    return counter

c1 = make_counter()
c2 = make_counter()

print(c1())
print(c1())
print(c2())

# 1
# 1
# 1
````

# Closures with `nonlocal` — Persistent State

```python
def make_counter():
    count = 0

    def counter():
        nonlocal count
        count += 1
        return count

    return counter


c1 = make_counter()

print(c1())
print(c1())

c2 = make_counter()

print(c2())
```

### Output

```text
1
2
1
```

### Why?

`c1 = make_counter()` creates **one closure** with its own:

```text
count = 0
```

Calling `c1()` twice modifies that **same `count`**:

```text
c1() → 1
c1() → 2
```

Then:

```python
c2 = make_counter()
```

calls `make_counter()` again, creating a **completely separate closure** with a new:

```text
count = 0
```

So:

```text
c2() → 1
```

###  Key Takeaway

```text
c1 → private count = 2
c2 → separate private count = 1
```

> **`nonlocal` allows the inner function to modify and preserve a variable from its enclosing function across multiple calls.**

Each separate call to `make_counter()` creates a **new independent closure and state**.

````python
def factorial(n):
    if n <= 1:
        return 1
    return n * factorial(n - 1)

print(factorial(5))

# 120
````

````python
nums = [1, 2, 3]
extra = {"sep": " - "}

def show(*args, sep=", "):
    print(sep.join(str(x) for x in args))

show(*nums, **extra)

# 1-2-3
````

````python
def f(x, y, z=3, *args, **kwargs):
    print(x, y, z, args, kwargs)

f(1, 2, 3, 4, 5, a=10, b=20)

# 1 2 3 4 5 a:10 b:20
````

````python
def f(a, b=[], c={}):
    b.append(a)
    c[a] = a
    return b, c

print(f(1))
print(f(2))

# [1] {1:1}
# [1,2] {1:1,2:2}
````
