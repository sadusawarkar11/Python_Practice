# NaN — `==` vs `is`

`NaN` (**Not a Number**) has a special behavior in Python and follows the IEEE 754 floating-point standard.

```python
x = float('nan')

print(x == x)
print(x is x)
```

### Output

```text
False
True
```

### Why?

#### `x == x` → `False`

`==` checks **value equality**.

But `NaN` is special:

```python
NaN == NaN
```

is always:

```text
False
```

So:

```python
x == x
```

→ **False**

#### `x is x` → `True`

`is` checks **object identity**, not value.

`x` is literally the same object as itself:

```python
x is x
```

→ **True**

###  Key Takeaway

```text
==  → checks values
is  → checks identity (same object)
```

For `NaN`:

```python
x == x    # False
x is x    # True
```

> **`NaN` is the classic example where `x == x` and `x is x` do not agree.**

# **Walrus Operator** `:=`

The **walrus operator (`:=`)** allows you to **assign a value to a variable as part of an expression**.

It was introduced in **Python 3.8**.

## Normal Way

```python
n = len(data)

if n > 10:
    print(f"Too long: {n}")
```

Here, assignment and checking happen in two separate steps.

## Using the Walrus Operator

```python
if (n := len(data)) > 10:
    print(f"Too long: {n}")
```

`n := len(data)` does **two things**:

1. Calculates `len(data)` and assigns it to `n`.
2. The expression evaluates to that same value, which is then compared with `10`.

So:

```text
(n := len(data))
      ↓
   assigns n
      ↓
   returns n
      ↓
   compares > 10
```

---

## Example with Regular Expressions

### Without `:=`

```python
import re

match = re.search(r'\d+', "abc123")

if match:
    print(match.group())
```

### With `:=`

```python
if match := re.search(r'\d+', "abc123"):
    print(match.group())
```

The result of `re.search()` is assigned to `match` and simultaneously used as the `if` condition.

---

## Example with `while`

```python
while (chunk := input("Enter text (or 'quit'): ")) != "quit":
    print(f"You entered: {chunk}")
```

Here, the input is:

1. Assigned to `chunk`
2. Compared with `"quit"`
3. Used inside the loop

---

##  Key Takeaway

```text
=   → assignment statement
:=  → assignment expression
```

Example:

```python
n := len(data)
```

means **assign the value to `n` and also use that value as part of the expression**.

### Why is it called "walrus"?

The `:=` symbol visually resembles a walrus's eyes and tusks when turned sideways. 🦭

> **Use `:=` when assigning and immediately using a value makes the code clearer and avoids repeating a computation.**

````python
data = [1, 2, 3, 4, 5]

if (n := len(data)) > 3:
    print(f"List has {n} items, which is more than 3")
else:
    print(f"List has {n} items")

# List has 5 items, which is more than 3
````

# Dictionary `in` Operator — Keys vs Values

```python
d = {"a": 1, "b": 2, "c": 3}

print(2 in d)
print(2 in d.values())
```

### Output

```text
False
True
```

### Why?

```python
2 in d
```

checks the **keys** of the dictionary.

Keys are:

```text
"a", "b", "c"
```

`2` is not a key → **False**

---

```python
2 in d.values()
```

checks the **values**:

```text
1, 2, 3
```

`2` is present → **True**

###  Key Takeaway

```python
x in d
```

→ checks **keys**

```python
x in d.values()
```

→ checks **values**

```python
x in d.keys()
```

→ explicitly checks **keys**

> **Remember: `in dictionary` searches keys by default, not values.**

````python
name = "Alice"

if "x" not in name:
    print("A")
else:
    print("B")

nums = range(1, 5)

if 4 in nums:
    print("C")
else:
    print("D")

if 5 in nums:
    print("E")
else:
    print("F")

# A
# C
# F
````

````python
nums = []

if all(n > 0 for n in nums):
    print("A")
else:
    print("B")

if any(n > 0 for n in nums):
    print("C")
else:
    print("D")


# A
# D
````

# `all()` vs `any()`

### 1. `all(iterable)`

Returns **`True` only if every element is truthy**.

```python
print(all([True, True, True]))    # True
print(all([True, False, True]))   # False
```

An **empty iterable** also returns `True`:

```python
print(all([]))   # True
```

This is called **vacuous truth**.

---

### 2. `any(iterable)`

Returns **`True` if at least one element is truthy**.

```python
print(any([False, False, True]))   # True
print(any([False, False, False]))  # False
```

An **empty iterable** returns `False`:

```python
print(any([]))   # False
```

###  Key Takeaway

```text
all() → Are ALL elements truthy?
any() → Is ANY element truthy?
```

| Function | Meaning | Empty Iterable |
|---|---|---|
| `all()` | Every element is truthy | `True` |
| `any()` | At least one element is truthy | `False` |

````python
def check(x):
    if x is None:
        return "none"
    if x < 0:
        return "negative"
    if x == 0:
        return "zero"
    return "positive"

print(check(None))
print(check(-5))
print(check(0))
print(check(10))

# none
# negative
# zero
# positive
````
