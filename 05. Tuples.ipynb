### `sorted()` with Tuples

`sorted()` **always returns a list**, no matter what iterable you pass to it.

### Example

```python
t = (5, 3, 8, 1, 9)

print(sorted(t))
# [1, 3, 5, 8, 9]

print(type(sorted(t)))
# <class 'list'>
```

---

### `sorted()` vs Slicing

**Slicing preserves the container type:**

```python
t[1:3]
# (3, 8)
```

A sliced tuple is still a **tuple**.

**`sorted()` does not preserve the container type:**

```python
sorted(t)
# [1, 3, 5, 8, 9]
```

It always returns a **list**.

---

### If You Need a Sorted Tuple

Convert the result back to a tuple:

```python
tuple(sorted(t))
# (1, 3, 5, 8, 9)
```

### Key Point

- `sorted()` → Always returns a **list**.
- Slicing (`[:]`) → Preserves the original container type.

**One-line summary:**
> `sorted()` always returns a **list**, even when the input is a tuple, string, set, or any other iterable.

### Hashable Tuples as Dictionary Keys

A tuple can be used as a dictionary key **only if every element inside it is hashable**.

### Example

```python
d2 = {}

d2[(1, [2, 3])] = "test"
```

Output:

```python
TypeError: unhashable type: 'list'
```

### Why?

The tuple:

```python
(1, [2, 3])
```

contains a **list**.

Lists are **mutable**, so they are **not hashable**.

Since one element is unhashable, the **entire tuple becomes unhashable**.

---

### Valid Tuple Keys

```python
d = {}

d[(1, 2)] = "OK"
d[("a", 3)] = "OK"
d[((1, 2), 3)] = "OK"
```

These work because every element is hashable.

---

### Invalid Tuple Keys

```python
(1, [2, 3])   # ❌ contains a list
(1, {2, 3})   # ❌ contains a set
(1, {"a": 1}) # ❌ contains a dictionary
```

Trying to use any of these as dictionary keys raises:

```python
TypeError
```

### Key Point

A tuple is hashable **only if all of its elements are hashable**.

**One-line summary:**
> Tuples can be dictionary keys only when every element inside them is hashable. If even one element is a list, dict, or set, Python raises `TypeError`.

````python
points = [(1, 2), (3, 4), (5, 6)]
for x, y in points:
    print(x + y)

total = sum(x * y for x, y in points)
print(total)

# 3
# 7
# 11
# 44
````

### Empty Tuple (`()`) is a Singleton

Normally, different tuple objects are different objects in memory.

**Exception:** The empty tuple `()` is a **singleton** in CPython.

### Example

```python
c = ()
d = ()

print(c is d)
# True
```

Both `c` and `d` refer to the **same empty tuple object**.

---

### Verify It

```python
print(() is ())
# True

print(id(()))
print(id(()))
# Same ID
```

The IDs are the same because Python reuses one shared empty tuple.

---

### Compare with Non-Empty Tuples

```python
a = (1, 2)
b = (1, 2)

print(a == b)
# True

print(a is b)
# Usually False
```

- `==` → Same values.
- `is` → Usually different objects.

> **Note:** Whether identical non-empty tuples are the same object can vary depending on Python optimizations. Do **not** rely on `is` for comparing tuples. Use `==` for value comparison.

### Key Point

- `()` → One shared singleton object.
- Non-empty tuples → Usually separate objects.

**One-line summary:**
> The empty tuple `()` is a singleton in CPython, so every `()` refers to the same object and `() is ()` is `True`.

````python
t = (1, 2)
original_id = id(t)
t += (3,)
print(t)
print(id(t) == original_id)

# (1,2,3)
# False
````

````python
def get_stats(nums):
    return min(nums), max(nums), sum(nums)

result = get_stats([3, 1, 4, 1, 5])
print(result)
print(type(result))

low, high, total = get_stats([3, 1, 4, 1, 5])
print(low, high, total)

# 1 5 14
# <class 'tuple'>
# 1 5 14
````

### Extended Unpacking (`*`)

When using `*` in unpacking, the starred variable **always becomes a list**, regardless of the source iterable.

### Example with a Tuple

```python
t = (1, 2, 3, 4, 5)

a, *b, c = t

print(a)       # 1
print(b)       # [2, 3, 4]
print(c)       # 5
print(type(b)) # <class 'list'>
```

---

### Example with a List

```python
lst = [1, 2, 3, 4, 5]

a, *b, c = lst

print(b)
# [2, 3, 4]
```

Even though the source is a list, `b` is still a **list**.

---

### Key Point

The starred variable (`*b`) **always collects the remaining items into a list**, no matter whether the source is:

- List ✅
- Tuple ✅
- String ✅
- Range ✅
- Any iterable ✅

### One-line Summary

> In extended unpacking, the `*` variable always receives a **list**, regardless of the type of the iterable being unpacked.
