# Python Variable Swapping

Python swaps variables using **tuple packing and unpacking**.

### Example

```python
a = 10
b = 20

a, b = b, a

print(a)  # 20
print(b)  # 10
```

---

## What Happens Internally?

Python first evaluates the **entire right-hand side** and creates a temporary tuple:

```python
(b, a)
```

Then it unpacks that tuple into the variables on the left:

```python
a, b = (b, a)
```

So the swap is effectively:

```python
temp = (b, a)   # Temporary tuple
a, b = temp     # Unpack into a and b
```

This is why no variable is overwritten during the swap.

---

## Why It Works

The **right-hand side is fully evaluated before any assignment happens**.

So Python first remembers the original values, then assigns them back in the new order.

---

## This Generalizes to More Variables

You can rotate multiple variables the same way:

```python
a, b, c = 1, 2, 3

a, b, c = c, a, b

print(a, b, c)
# 3 1 2
```

Python internally treats this like:

```python
(a, b, c) = (c, a, b)
```

---

## Key Point

Python's variable swap is **not magic**.

It uses a **temporary tuple** behind the scenes:

1. Pack values into a temporary tuple.
2. Unpack the tuple into the target variables.

## One-Line Summary

> `a, b = b, a` works because Python first creates a temporary tuple `(b, a)` and then unpacks it into `a` and `b`.

# `bool` is a Subclass of `int`

In Python, `bool` inherits from `int`.

```python
print(bool.__bases__)
# (<class 'int'>,)

print(isinstance(True, int))
# True
```

This means:

- `True` behaves like the integer `1`.
- `False` behaves like the integer `0`.

However, `bool` is still its **own type**.

---

## 1. Booleans Work in Arithmetic

```python
print(True + True)
# 2

print(True + 1)
# 2

print(False * 10)
# 0
```

A very useful trick:

```python
print(sum([True, True, False, True]))
# 3
```

`sum()` counts the number of `True` values.

This is commonly used in **NumPy** and **Pandas**:

```python
sum(df["age"] > 30)
```

Counts how many rows satisfy the condition.

---

## 2. Booleans Can Be Used as List Indices

```python
lst = ["a", "b", "c"]

print(lst[True])
# 'b'

print(lst[False])
# 'a'
```

Why?

```python
True  == 1
False == 0
```

So:

- `lst[True]` → `lst[1]`
- `lst[False]` → `lst[0]`

⚠️ This can cause subtle bugs if a function accidentally receives a boolean instead of an integer index.

---

## 3. `True` and `1` Collide in Dictionaries

```python
d = {True: "yes", 1: "no"}

print(d)
# {True: 'no'}
```

Why?

```python
True == 1
# True
```

They also have the same hash value, so Python treats them as the **same dictionary key**.

The second assignment overwrites the first.

The same applies to:

```python
False == 0
```

---

## 4. `type()` Still Distinguishes Them

```python
print(type(True))
# <class 'bool'>

print(type(1))
# <class 'int'>
```

Although:

```python
print(True == 1)
# True
```

their types are different:

```python
print(type(True) == type(1))
# False
```

---

## `==` vs `is`

```python
print(True == 1)
# True
```

`==` compares **values**.

```python
print(True is 1)
# False
```

`is` compares **object identity**.

> **Note:** In modern Python (3.8+), `True is 1` produces `False`, but it also raises a `SyntaxWarning` because comparing a literal with `is` is discouraged. Prefer `==` for value comparisons.

---

## Key Points

- `bool` is a subclass of `int`.
- `True` behaves like `1`.
- `False` behaves like `0`.
- Booleans work in arithmetic.
- `sum()` can count `True` values.
- Booleans can be used as list indices (`True → 1`, `False → 0`).
- `True` and `1` (and `False` and `0`) are equal, so they collide as dictionary keys.
- `type(True)` is still `bool`, not `int`.

## One-Line Summary

> `bool` is a subclass of `int`, so `True` behaves like `1` and `False` like `0`, but `bool` remains a distinct type with its own identity.

# Comparing Different Types with `==`

### Example

```python
x = 10
y = "10"

print(x == y)
# False
```

---

## Why Does It Return `False`?

`x` is an **integer** and `y` is a **string**.

Python does **not** automatically convert one type into another when using `==`.

Instead, it tries to compare them.

---

## What Happens Internally?

When Python evaluates:

```python
x == y
```

it performs these steps:

1. Calls:

```python
x.__eq__(y)
```

which is:

```python
int.__eq__(10, "10")
```

Since `int` doesn't know how to compare itself with a `str`, it returns:

```python
NotImplemented
```

2. Python then tries the reverse comparison:

```python
y.__eq__(x)
```

which is:

```python
str.__eq__("10", 10)
```

`str` also doesn't know how to compare itself with an `int`, so it also returns:

```python
NotImplemented
```

3. Since both comparisons return `NotImplemented`, Python falls back to:

```python
False
```

No exception is raised.

---

## Compare After Conversion

```python
int("10") == 10
# True
```

or

```python
str(10) == "10"
# True
```

Now both operands have the same type, so the values compare as equal.

---

## Key Point

- `==` compares values.
- Python does **not** automatically convert between `int` and `str`.
- If both objects cannot compare themselves to each other (both return `NotImplemented`), Python returns `False`.

## One-Line Summary

> When comparing an `int` and a `str`, Python first tries `int.__eq__()`, then `str.__eq__()`. If both return `NotImplemented`, Python returns `False` instead of raising an error.

# Assignment Creates a Reference, Not a Copy

## Rule 1: Assignment Does **Not** Make a Copy

When you write:

```python
a = 5
b = a
```

`b` does **not** get a copy of `a`. Both variables point to the **same object**.

```python
print(id(a) == id(b))
# True
```

This is true for **all Python objects** (integers, strings, lists, tuples, dictionaries, etc.).

---

## Rule 2: Immediately After `b = a`

Immediately after:

```python
b = a
```

both variables refer to the **same object**.

```python
print(a is b)
# True

print(id(a) == id(b))
# True
```

### Examples

```python
# Integer
a = 1000
b = a
print(a is b)      # True
```

```python
# String
a = "Python"
b = a
print(a is b)      # True
```

```python
# List
a = [1, 2, 3]
b = a
print(a is b)      # True
```

This happens because **assignment creates another reference, not a copy**.

---

## Rule 3: What Happens After Reassignment?

### Immutable Objects (`int`, `str`, `tuple`, etc.)

```python
a = 5
b = a

b = b + 1

print(a)      # 5
print(b)      # 6
print(a is b) # False
```

Why?

- Integers are **immutable**.
- `b = b + 1` cannot modify `5`.
- Python computes the value `6`.
- `b` is rebound to the object representing `6`.
- `a` still points to `5`.

Whether `6` comes from Python's small-integer cache (`-5` to `256`) or is newly created **does not change the behavior**.

---

### Mutable Objects (`list`, `dict`, `set`, etc.)

```python
a = [1, 2, 3]
b = a

b.append(4)

print(a)
# [1, 2, 3, 4]

print(b)
# [1, 2, 3, 4]

print(a is b)
# True
```

Lists are **mutable**, so changing the object through one variable changes what both variables see.

---

## Visual Representation

Initially:

```text
a ─────► 5
b ─────► 5
```

After:

```python
b = b + 1
```

It becomes:

```text
a ─────► 5

b ─────► 6
```

`a` never changes.

---

## Key Points

- `b = a` → Creates **another reference**, **not a copy**.
- Immediately after `b = a`, `a is b` is **always `True`**.
- Immutable objects create a new binding when "modified".
- Mutable objects are modified **in place**, so all references see the change.

## One-Line Summary

> **`b = a` never copies an object—it simply makes `b` point to the same object as `a`. Immediately after assignment, `a is b` (and `id(a) == id(b)`) is always `True`.**

# List Aliasing

Assigning one list to another variable **does not create a copy**.

Both variables point to the **same list object**.

### Example

```python
original = [1, 2, 3]
copy = original

copy.append(4)

print(original)
# [1, 2, 3, 4]

print(copy)
# [1, 2, 3, 4]

print(original is copy)
# True
```

---

## Why?

```python
copy = original
```

does **not** create a new list.

It simply creates another **reference** to the same list.

When you do:

```python
copy.append(4)
```

the list itself is modified **in place**.

Since both variables point to the same object, both see the change.

---

## Visual Representation

Initially:

```text
original ─────► [1, 2, 3]
copy     ─────► [1, 2, 3]
```

After:

```python
copy.append(4)
```

Both variables still point to the same list:

```text
original ─────► [1, 2, 3, 4]
copy     ─────► [1, 2, 3, 4]
```

---

## If You Want an Independent Copy

Use `.copy()`:

```python
original = [1, 2, 3]
copy = original.copy()

copy.append(4)

print(original)
# [1, 2, 3]

print(copy)
# [1, 2, 3, 4]

print(original is copy)
# False
```

---

## Key Point

- `copy = original` → Creates an **alias**, not a copy.
- `.append()` **mutates** the existing list.
- Both variables see the change because they reference the same object.

## One-Line Summary

> `copy = original` creates another reference to the same list, so modifying the list through one variable (e.g., using `.append()`) changes what both variables see.

# `UnboundLocalError` in Python

### Example

```python
a = 5

def foo():
    print(a)
    a = 10

foo()
```

Output:

```python
UnboundLocalError: local variable 'a' referenced before assignment
```

---

## Why Doesn't It Print `5`?

You might expect:

- `print(a)` runs first.
- So it should print the global value `5`.

**But that's not how Python works.**

---

## Python Decides Scope Before Execution

Before the function runs, Python scans the **entire function body**.

It asks:

> **"Is `a` assigned anywhere in this function?"**

It finds:

```python
a = 10
```

So Python decides:

> **`a` is a local variable for the entire function.**

This decision is made **before** the function starts executing.

---

## What Happens When `foo()` Runs?

Python has already decided that `a` is local.

Execution begins:

```python
print(a)
```

Python looks for the **local** variable `a`.

But the assignment:

```python
a = 10
```

has **not executed yet**, so the local variable has no value.

Result:

```python
UnboundLocalError
```

Python **does not** fall back to the global variable `a = 5`.

---

## Visual Flow

```text
Global:
a = 5

Function foo():

print(a)   ← Python looks for LOCAL a
a = 10     ← Assignment makes a LOCAL for the entire function
```

Since the local `a` hasn't been assigned when `print(a)` executes:

```text
UnboundLocalError
```

---

## How to Use the Global Variable

If you want to modify the global variable, declare it with `global`:

```python
a = 5

def foo():
    global a
    print(a)
    a = 10

foo()

print(a)
# 10
```

---

## Key Point

- Python decides whether a variable is **local or global before the function runs**.
- If a variable is assigned anywhere in a function, Python treats it as **local throughout the entire function**.
- Accessing that local variable before it has been assigned raises `UnboundLocalError`.

## One-Line Summary

> Python determines variable scope at compile time. If a variable is assigned anywhere in a function, it is treated as local throughout the function, so accessing it before the assignment raises `UnboundLocalError`.

# `+=` and `UnboundLocalError`

### Example

```python
count = 0

def increment():
    count += 1
    print(count)

increment()
```

Output:

```python
UnboundLocalError: cannot access local variable 'count' where it is not associated with a value
```

---

## Why Does This Happen?

The statement:

```python
count += 1
```

is equivalent to:

```python
count = count + 1
```

So Python sees an **assignment** to `count`.

Before the function runs, Python scans the entire function and decides:

> **`count` is a local variable for the entire function.**

---

## What Happens During Execution?

Python treats `count += 1` as:

```python
count = count + 1
```

To evaluate the right-hand side:

```python
count + 1
```

Python must first read the current value of the **local** variable `count`.

But the local `count` has **not been assigned yet**.

Result:

```python
UnboundLocalError
```

Python **does not** use the global variable `count = 0`.

---

## Visual Flow

```text
Global:
count = 0

Function increment():

count += 1
│
├── Equivalent to:
│   count = count + 1
│
├── Python already decided:
│   count is LOCAL
│
├── Read local count
│   ❌ No value yet
│
└── UnboundLocalError
```

---

## Correct Way (Using `global`)

```python
count = 0

def increment():
    global count
    count += 1
    print(count)

increment()
# 1
```

Now Python knows `count` refers to the global variable.

---

## Key Point

- `count += 1` is equivalent to `count = count + 1`.
- Because there is an assignment, Python treats `count` as **local** for the entire function.
- Reading that local variable before it has a value raises `UnboundLocalError`.

## One-Line Summary

> `+=` is an assignment operation. If a variable is assigned anywhere in a function, Python treats it as local throughout the function, so `count += 1` raises `UnboundLocalError` unless the variable is declared `global` (or `nonlocal` in nested functions).

# Scope Analysis in Nested Functions

### Example

```python
x = 100

def outer():
    print(x)

    def inner():
        print(x)
        x = 5

    inner()

outer()
```

Output:

```text
100

Traceback (most recent call last):
...
UnboundLocalError: cannot access local variable 'x' where it is not associated with a value
```

---

## Step 1: Analyze `outer()`

Python looks only at the body of `outer()`.

```python
def outer():
    print(x)

    def inner():
        print(x)
        x = 5

    inner()
```

Does `outer()` assign to `x` anywhere in **its own body**?

**No.**

The assignment:

```python
x = 5
```

belongs to **`inner()`**, not `outer()`.

So inside `outer()`, `x` refers to the **global** variable.

Therefore:

```python
print(x)
```

prints:

```text
100
```

---

## Step 2: Analyze `inner()`

Now Python analyzes `inner()` separately.

```python
def inner():
    print(x)
    x = 5
```

It finds:

```python
x = 5
```

So Python decides:

> **`x` is a local variable throughout `inner()`.**

When execution reaches:

```python
print(x)
```

Python looks for the **local** `x`.

But it has not been assigned yet.

Result:

```text
UnboundLocalError
```

Python does **not** use:

- the global `x`
- or the `x` visible from `outer()`

because `x` has already been determined to be local to `inner()`.

---

## Scope Analysis Is Per Function

Python analyzes **each function independently**.

### `outer()`

```text
Assignments to x?
❌ No

→ x is global
```

### `inner()`

```text
Assignments to x?
✅ Yes

→ x is local
```

The assignment inside `inner()` **does not affect** how Python treats `x` inside `outer()`.

---

## Visual Representation

```text
Global Scope
│
├── x = 100
│
└── outer()
    │
    ├── print(x) ─────► Uses GLOBAL x (100)
    │
    └── inner()
         │
         ├── print(x)
         │      ↓
         │   Looks for LOCAL x
         │      ❌ Not assigned yet
         │
         └── x = 5
```

Result:

```text
100
UnboundLocalError
```

---

## Key Point

- Python determines variable scope **separately for each function**.
- Assignments inside a nested function affect **only that function's scope**.
- An assignment inside `inner()` does **not** make `x` local in `outer()`.

## One-Line Summary

> Python analyzes each function independently. `outer()` uses the global `x`, while `inner()` treats `x` as local because it assigns to it, causing `UnboundLocalError` when `print(x)` executes first.

````python
a = 5
def foo():
    b = a + 1   # 'a' is never assigned inside foo(), so this reads the global a
    print(b)
foo()
````

# `UnboundLocalError` in Nested Functions

### Example

```python
def foo():
    a = 1

    def bar():
        a = a + 1
        return a

    return bar()

foo()
```

Output:

```text
UnboundLocalError: cannot access local variable 'a' where it is not associated with a value
```

---

## Why Does This Happen?

Look at the body of `bar()`:

```python
def bar():
    a = a + 1
    return a
```

Python sees:

```python
a = ...
```

Since `a` is assigned inside `bar()`, Python decides:

> **`a` is a local variable throughout `bar()`.**

This decision is made **before** `bar()` runs.

---

## What Happens During Execution?

The statement:

```python
a = a + 1
```

is evaluated as:

1. Read the current value of **local** `a`.
2. Add `1`.
3. Store the result back into **local** `a`.

But when Python tries to read:

```python
a + 1
```

the local `a` has **not been assigned yet**.

Result:

```text
UnboundLocalError
```

---

## Why Doesn't It Use `foo()`'s `a`?

Although `foo()` has:

```python
a = 1
```

that does **not** matter.

Python has already decided that `a` inside `bar()` is **local**, so it **does not** look in the enclosing function.

Assignments inside a function always make that name local (unless you use `global` or `nonlocal`).

---

## Visual Representation

```text
foo()
│
├── a = 1
│
└── bar()
     │
     ├── a = a + 1
     │
     ├── Python already decided:
     │   a is LOCAL
     │
     ├── Read local a
     │      ❌ No value yet
     │
     └── UnboundLocalError
```

---

## How to Modify the Enclosing Variable

Use `nonlocal`:

```python
def foo():
    a = 1

    def bar():
        nonlocal a
        a = a + 1
        return a

    return bar()

print(foo())
# 2
```

`nonlocal` tells Python to use the variable from the nearest enclosing function instead of creating a new local one.

---

## Key Point

- Python analyzes each function separately.
- An assignment inside `bar()` makes `a` local to `bar()`.
- Python does **not** automatically use the variable from the enclosing function.
- To modify an enclosing function's variable, use `nonlocal`.

## One-Line Summary

> Because `bar()` assigns to `a`, Python treats `a` as local throughout `bar()`. Reading it before assignment causes `UnboundLocalError`; use `nonlocal` to access and modify the enclosing function's variable.

````python
a = [1, 2, 3]
b = a
print(id(a) == id(b))   # True

a = a + [4]
print(a)   # [1, 2, 3, 4]
print(b)   # [1, 2, 3]  ← unchanged!
print(id(a) == id(b))   # False — a now points to a new object, b still points to the old one
````

````python
a = (1, 2, 3)
b = a
print(id(a) == id(b))   # True — same object, as always right after b = a

b += (4,)
# no __iadd__ exists for tuples → falls back to b = b + (4,)
# → creates a NEW tuple (1,2,3,4), rebinds b to it, a is untouched

print(a)                # (1, 2, 3)       — unchanged
print(b)                # (1, 2, 3, 4)    — new object
print(id(a) == id(b))   # False
````

# List Aliasing Inside a Function

### Example

```python
def double_numbers(numbers):
    result = numbers

    for i in range(len(result)):
        result[i] = result[i] * 2

    return result


original = [1, 2, 3]
doubled = double_numbers(original)

print(original)
# [2, 4, 6]

print(doubled)
# [2, 4, 6]

print(original is doubled)
# True
```

---

## Why Does `original` Change?

The statement:

```python
result = numbers
```

does **not** create a copy.

It simply creates another **reference** to the same list object.

So:

```text
original ───► [1, 2, 3]
numbers  ───► [1, 2, 3]
result   ───► [1, 2, 3]
```

All three names point to the **same list**.

---

## What Happens in the Loop?

```python
result[i] = result[i] * 2
```

This modifies the existing list **in place**.

Example:

```python
result[0] = result[0] * 2
```

changes:

```text
[1, 2, 3]
```

to

```text
[2, 2, 3]
```

Since `original`, `numbers`, and `result` all reference the **same object**, they all see the change.

---

## Visual Representation

Before the loop:

```text
original ──┐
numbers  ──┼────► [1, 2, 3]
result   ──┘
```

After the loop:

```text
original ──┐
numbers  ──┼────► [2, 4, 6]
result   ──┘
```

Only **one list object** exists.

---

## How to Avoid Modifying the Original List

Create a copy first:

```python
def double_numbers(numbers):
    result = numbers.copy()

    for i in range(len(result)):
        result[i] *= 2

    return result
```

Now:

```python
original = [1, 2, 3]
doubled = double_numbers(original)

print(original)
# [1, 2, 3]

print(doubled)
# [2, 4, 6]

print(original is doubled)
# False
```

---

## Key Point

- `result = numbers` creates an **alias**, not a copy.
- Both variables point to the **same list object**.
- Modifying `result` modifies `numbers` (and the caller's `original`) because they all reference the same object.

## One-Line Summary

> `result = numbers` does not copy the list—it creates another reference to the same object, so in-place changes through `result` also modify the caller's original list.
