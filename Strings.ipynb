### `split()` vs `split(" ")`

These two methods **do not** behave the same.

### 1. `split()` (No Arguments)

- Splits on **any whitespace** (spaces, tabs, newlines).
- Ignores extra, leading, and trailing whitespace.
- Does **not** return empty strings.

```python
s = "  the quick   brown fox  "

print(s.split())
# ['the', 'quick', 'brown', 'fox']
```

### 2. `split(" ")` (Explicit Space)

- Splits only on the **space character** (`" "`).
- Does **not** ignore extra spaces.
- Returns empty strings (`''`) where there are consecutive, leading, or trailing spaces.

```python
print(s.split(" "))
# ['', '', 'the', 'quick', '', '', 'brown', 'fox', '', '']
```

### Key Point

- `split()` → **Smart** whitespace splitting (recommended in most cases).
- `split(" ")` → **Literal** splitting on spaces (keeps empty strings).

**One-line summary:**
> Use `split()` to ignore extra spaces. Use `split(" ")` only when you want to split on every single space and keep empty results.

````python
s = "Data Science"
print(s.startswith("Data"))
print(s.endswith("science"))
print(s.endswith("Science"))

# True
# False
# True
````

### `split()` and `join()`

`split()` and `join()` are opposite operations.

### `split()`

- Takes **one string** as input.
- Returns a **list of strings**.

```python
s = "a-b-c-d"

parts = s.split("-")
print(parts)
# ['a', 'b', 'c', 'd']
```

Since there is only **one `-` between each letter**, there are **no empty strings**.

### `join()`

- Takes a **list (or iterable) of strings**.
- Returns **one single string**.

```python
joined = "_".join(parts)
print(joined)
# 'a_b_c_d'
```

`join()` **never** returns a list.

### `join()` with Numbers

```python
nums = [1, 2, 3]

"-".join(nums)
```

Output:

```python
TypeError: sequence item 0: expected str instance, int found
```

`join()` only works with **strings**.

Convert numbers first:

```python
"-".join(map(str, nums))
# '1-2-3'
```

or

```python
"-".join(str(n) for n in nums)
# '1-2-3'
```

### Key Point

- `split()` → **String → List of strings**
- `join()` → **List of strings → String**

**One-line summary:**
> `split()` breaks one string into a list, while `join()` combines a list of strings into one string. `join()` does not automatically convert numbers to strings.

### Python f-String Format Specifiers

General syntax:

```python
{value:alignment width.precision type}
```

You only need the parts you want.

---

### 1. Alignment + Width

- `<` → Left align
- `>` → Right align
- `^` → Center align

Example:

```python
print(f"{'Name':<10}{'Age':>5}")
```

Output:

```text
Name        Age
```

- `'Name':<10` → `"Name"` + 6 spaces
- `'Age':>5` → 2 spaces + `"Age"`

---

### 2. Float Precision (`.2f`)

Shows **2 digits after the decimal point**.

```python
print(f"{3.14159:.2f}")
# 3.14
```

---

### 3. Zero Padding (`05d`)

- `d` → Integer
- `5` → Total width is 5
- `0` → Pad with zeros

```python
print(f"{42:05d}")
# 00042
```

Since `42` has 2 digits, Python adds **3 leading zeros** to make the width 5.

---

### Repeating Characters

```python
print("=" * 20)
```

Output:

```text
====================
```

(Exactly **20** `=` characters.)

### Key Point

- `<`, `>`, `^` → Alignment
- `width` → Total field width
- `.2f` → 2 decimal places
- `05d` → Pad an integer with leading zeros

**One-line summary:**
> f-strings let you control alignment, width, decimal places, and zero-padding using `{value:format_spec}`.

### `==` vs `is` (String Interning)

- `==` → Checks if **values are equal**.
- `is` → Checks if both variables refer to the **same object in memory**.

---

### Example 1

```python
a = "hello"
b = "hello"

print(a is b)
# True
```

`"hello"` is **interned** by Python, so `a` and `b` point to the same object.

---

### Example 2

```python
c = "hello world"
d = "hello world"

print(c is d)
# False
```

Strings with **spaces** are usually **not interned**, so they are different objects.

---

### Example 3

```python
e = "hello_world"
f = "hello_world"

print(e is f)
# True
```

`"hello_world"` is a **valid Python identifier** (letters + underscore), so Python interns it.

### Key Point

Python commonly interns strings that look like **valid identifiers** (letters, digits, and underscores, **no spaces**).

- `"hello"` → Interned ✅
- `"hello_world"` → Interned ✅
- `"hello world"` → Usually not interned ❌

> **Note:** String interning is an implementation detail (common in CPython). Do **not** rely on `is` to compare strings. Always use `==` to compare string values.

**One-line summary:**
> Use `==` to compare string values. `is` checks object identity, and some identifier-like strings are automatically interned by Python.

### String Indexing vs Slicing

**Important Rule:**

- **Indexing (`s[i]`)** → Raises `IndexError` if the index is out of range.
- **Slicing (`s[start:end]`)** → **Never** raises an error for out-of-range indices. Python automatically adjusts (clamps) the indices.

---

### Example

```python
s = "python"

print(s[100:200])   # ''
print(s[2:100])     # 'thon'
print(s[-100:3])    # 'pyt'
```

Why?

- `s[100:200]` → Range is beyond the string → returns `''` (empty string).
- `s[2:100]` → End is beyond the string → treated as `s[2:]` → `'thon'`.
- `s[-100:3]` → Start is before the beginning → treated as `s[0:3]` → `'pyt'`.

---

### Indexing is Different

```python
s[100]
```

Output:

```python
IndexError: string index out of range
```

This happens because indexing asks for **one specific character**, while slicing asks for **a range**.

### Key Point

- `s[100]` → ❌ `IndexError`
- `s[100:200]` → ✅ `''`
- `s[-100:3]` → ✅ `'pyt'`

**One-line summary:**
> **Indexing is strict** (out of range = error), but **slicing is forgiving** (out-of-range indices are automatically adjusted).

### String Repetition (`*`)

The `*` operator repeats a string.

### Examples

```python
print("abc" * 3)
# 'abcabcabc'
```

```python
print("abc" * 0)
# ''
```

```python
print("abc" * -5)
# ''
```

Any **zero or negative** repeat count returns an **empty string**.

---

### Type

```python
print(type("abc" * 0))
# <class 'str'>
```

Even an empty result is still a **string**.

---

### Don't Confuse These

**Negative multiplication (`*`)**

```python
"abc" * -5
# ''
```

→ Returns an empty string.

**Negative slicing (`[::-1]`)**

```python
"abc"[::-1]
# 'cba'
```

→ Reverses the string.

### Key Point

- `* 0` → Empty string (`''`)
- `* negative` → Empty string (`''`)
- `[::-1]` → Reverses the string

**One-line summary:**
> A negative repeat count does **not** reverse a string—it behaves like `0` and returns an empty string.

````python
s = "hello world"
print("hello" in s)
print("Hello" in s)
print("" in s)
print(["h", "e"] in s)

# True
# False
# False
# True
````

### `in` Operator with Strings

The `in` operator checks whether a **substring** exists inside another string.

### Examples

```python
s = "hello world"

print("hello" in s)   # True
print("Hello" in s)   # False
```

String comparisons are **case-sensitive**.

---

### Empty String

```python
print("" in s)
# True
```

An empty string (`""`) is considered to be present in **every** string.

---

### Wrong Type

```python
print(["h", "e"] in s)
```

Output:

```python
TypeError: 'in <string>' requires string as left operand, not list
```

When checking inside a string, the left operand **must also be a string**.

---

### If You Want to Check Multiple Characters

```python
all(char in s for char in ["h", "e"])
# True
```

This checks each character separately.

### Key Point

- `"hello" in s` → ✅ Checks for a substring.
- `"" in s` → ✅ Always `True`.
- `["h", "e"] in s` → ❌ `TypeError` (left operand must be a string).

**One-line summary:**
> When using `in` with a string, the left side must be a string. An empty string is always found, but a list causes a `TypeError`.
