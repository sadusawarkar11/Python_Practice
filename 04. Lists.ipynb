### `list.sort()` vs `sorted()`

These two are **not** the same.

### `list.sort()`

- Sorts the **original list** (in-place).
- Returns **`None`**.

```python
lst = [3, 1, 4, 1, 5, 9]

result = lst.sort()

print(result)   # None
print(lst)      # [1, 1, 3, 4, 5, 9]
```

---

### `sorted()`

- Creates a **new sorted list**.
- Leaves the original list unchanged.

```python
lst2 = [3, 1, 4, 1, 5, 9]

result2 = sorted(lst2)

print(result2)  # [1, 1, 3, 4, 5, 9]
print(lst2)     # [3, 1, 4, 1, 5, 9]
```

---

### Key Point

- `list.sort()` → **Changes the original list** and returns **`None`**.
- `sorted()` → **Returns a new sorted list** and does **not** change the original.

### Common Beginner Mistake

```python
x = lst.sort()

print(x)
# None
```

If you want the sorted list, use:

```python
x = sorted(lst)
```

or

```python
lst.sort()   # Then use lst
```

### Remember

Most list methods that **modify the list** return `None`, for example:

- `sort()`
- `reverse()`
- `append()`
- `extend()`
- `insert()`
- `remove()`

**Exception:** `pop()` modifies the list **and** returns the removed element.

**One-line summary:**
> `sort()` modifies the original list and returns `None`, while `sorted()` returns a new sorted list and leaves the original unchanged.

````python
lst = [10, 20, 30, 40, 50]
lst.remove(30)
print(lst)

lst2 = [10, 20, 30, 40, 50]
popped = lst2.pop(2)
print(lst2)
print(popped)

lst3 = [10, 20, 30, 40, 50]
del lst3[2]
print(lst3)

# [10,20,40,50]
# [10,20,40,50]
# 30
# [10,20,40,50]
````

### Passing Lists to Functions

Lists are **mutable**, and Python passes the **reference** to the function (it does **not** make a copy).

### Example

```python
def add_item(lst, item):
    lst.append(item)
    return lst

original = [1, 2, 3]

result = add_item(original, 4)

print(original)           # [1, 2, 3, 4]
print(result)             # [1, 2, 3, 4]
print(original is result) # True
```

### Why is `original is result` `True`?

- `lst` and `original` refer to the **same list**.
- `append()` modifies that same list.
- `return lst` returns the **same object**, not a copy.

So:

- `original`
- `lst` (inside the function)
- `result`

all refer to **one list object**.

---

### If You Want to Keep the Original List Unchanged

Make a copy first:

```python
def add_item_safe(lst, item):
    new_lst = lst.copy()   # or lst[:] or list(lst)
    new_lst.append(item)
    return new_lst
```

Now:

```python
original = [1, 2, 3]
result = add_item_safe(original, 4)

print(original)           # [1, 2, 3]
print(result)             # [1, 2, 3, 4]
print(original is result) # False
```

### Key Point

- Passing a list to a function **does not create a copy**.
- Changes made inside the function affect the original list.
- Use `.copy()`, `lst[:]`, or `list(lst)` if you need an independent list.

**One-line summary:**
> Lists are passed by reference (object reference), so modifying a list inside a function also modifies the original list unless you make a copy first.

### Creating a 2D List: Avoid `*` with Nested Lists

### Wrong Way

```python
matrix = [[0] * 3] * 3

matrix[0][0] = 1

print(matrix)
# [[1, 0, 0], [1, 0, 0], [1, 0, 0]]
```

**Why?**

`[[0] * 3] * 3` does **not** create 3 separate rows.

It creates **one inner list** and repeats its **reference** three times.

So all rows point to the **same list**.

---

### Correct Way

Use a list comprehension:

```python
matrix = [[0] * 3 for _ in range(3)]

matrix[0][0] = 1

print(matrix)
# [[1, 0, 0], [0, 0, 0], [0, 0, 0]]
```

A new row is created on each iteration, so each row is independent.

---

### Key Point

❌ Wrong:

```python
[[0] * 3] * 3
```

→ One row repeated by reference.

✅ Correct:

```python
[[0] * 3 for _ in range(3)]
```

→ Three independent rows.

**One-line summary:**
> Never use `[[...]] * n` to create a 2D list with mutable rows. Use a list comprehension so each row is a separate list.

````python
result = [x * y for x in range(1, 4) for y in range(1, 4)]
print(result)

pairs = [(x, y) for x in range(3) for y in range(3) if x != y]
print(pairs)

# [1,2,3,2,4,6,3,6,9]
# [(0,1),(0,2),(1,0),(1,2),(2,0),(2,1)]
````

### `==` vs `is` (Lists)

- `==` → Checks if **values are equal**.
- `is` → Checks if both variables refer to the **same object**.

### Example

```python
a = [1, 2, 3]
b = [1, 2, 3]
c = a

print(a == b)   # True
print(a == c)   # True
print(a is b)   # False
print(a is c)   # True
```

### Why?

- `a` and `b` are created from **two separate list literals**, so they are different objects.
- `c = a` does **not** create a copy. It simply makes `c` another name for the same list.

So:

- `a == b` → `True` (same values)
- `a is b` → `False` (different objects)
- `a is c` → `True` (same object)

### Key Point

```python
c = a
```

This creates an **alias**, not a copy.

If you need a copy:

```python
c = a.copy()
# or
c = a[:]
```

Now:

```python
a is c
# False
```

### Remember

- `==` → Compare **values**
- `is` → Compare **object identity (memory reference)**

**One-line summary:**
> `c = a` does not copy the list—it creates another reference to the same object, so `a is c` is `True`.

````python
def add_item(item, target=None):
    if target is None:
        target = []
    target.append(item)
    return target

print(add_item(1))
print(add_item(2))
print(add_item(3))

# [1]
# [2]
# [3]
````

### `+=` vs `+` Inside Functions

Lists are **mutable**, so `+=` and `+` behave differently.

### Example

```python
def modify_a(lst):
    lst += [4]

def modify_b(lst):
    lst = lst + [4]

a = [1, 2, 3]
modify_a(a)
print(a)
# [1, 2, 3, 4]

b = [1, 2, 3]
modify_b(b)
print(b)
# [1, 2, 3]
```

---

### `+=` (In-place Modification)

```python
lst += [4]
```

- Modifies the **same list**.
- The caller's list also changes.

So:

```python
a
# [1, 2, 3, 4]
```

---

### `+` (Creates a New List)

```python
lst = lst + [4]
```

- Creates a **new list**.
- Only the local variable `lst` points to the new list.
- The original list is unchanged.

So:

```python
b
# [1, 2, 3]
```

---

### Key Point

- `+=` → **Modifies the existing list**.
- `+` → **Creates a new list** and reassigns the local variable.

### Remember

```python
lst += [4]      # Changes the original list

lst = lst + [4] # Creates a new list (original unchanged)
```

**One-line summary:**
> `+=` mutates the original list, while `+` creates a new list. Inside a function, rebinding with `=` affects only the local variable, not the caller's list.

````python
lst = [1, 2, [3, 4]]
print(3 in lst)
print([3, 4] in lst)
print(3 in lst[2])

# False
# True
# False
````

### `in` with Nested Lists

The `in` operator checks **only the current list**, not nested lists automatically.

### Example

```python
lst = [1, 2, [3, 4]]

print(3 in lst)       # False
print([3, 4] in lst)  # True
print(3 in lst[2])    # True
```

### Why?

```python
3 in lst
# False
```

The outer list contains:

```python
[1, 2, [3, 4]]
```

Its elements are:

- `1`
- `2`
- `[3, 4]`

There is **no top-level element `3`**, so the result is `False`.

---

```python
[3, 4] in lst
# True
```

The inner list `[3, 4]` is a **top-level element** of `lst`.

---

```python
3 in lst[2]
# True
```

`lst[2]` is:

```python
[3, 4]
```

Now you're checking:

```python
3 in [3, 4]
```

Since `3` is an element of that inner list, the result is `True`.

### Key Point

- `3 in lst` → Checks only the **outer list**.
- `3 in lst[2]` → Checks the **inner list**.

**One-line summary:**
> The `in` operator does not search nested lists automatically. Access the inner list first (e.g., `lst[2]`) before checking its elements.

````python
lst = []
if lst:
    print("has items")
else:
    print("empty")

lst2 = [0]
if lst2:
    print("has items")
else:
    print("empty")

lst3 = [[]]
if lst3:
    print("has items")
else:
    print("empty")

# empty
# has items
# empty
````

### Truthiness of Nested Lists

For lists, **truthiness depends only on whether the outer list is empty**.

### Example

```python
lst3 = [[]]

if lst3:
    print("has items")
else:
    print("empty")
```

Output:

```text
has items
```

### Why?

`lst3` is:

```python
[[]]
```

The outer list contains **one element** (an empty list), so its length is `1`.

```python
len(lst3)
# 1
```

A non-empty list is **truthy**, even if its elements are empty or falsy.

### Compare

```python
bool([])      # False
bool([[]])    # True
bool([0])     # True
bool([""])    # True
```

- `[]` → Empty list → `False`
- `[[]]` → One element (an empty list) → `True`
- `[0]` → One element (`0`) → `True`
- `[""]` → One element (empty string) → `True`

### Key Point

Python checks **only the outer container**, not the truthiness of its contents.

**One-line summary:**
> `[]` is falsy, but `[[]]` is truthy because the outer list contains one element, even though that element is an empty list.
