# NaN — `==` vs `is`

`NaN` (**Not a Number**) has a special behavior in Python and follows the IEEE 754 floating-point standard.

```python
x = float('nan')

print(x == x)
print(x is x)
```

### Output

```text
False
True
```

### Why?

#### `x == x` → `False`

`==` checks **value equality**.

But `NaN` is special:

```python
NaN == NaN
```

is always:

```text
False
```

So:

```python
x == x
```

→ **False**

#### `x is x` → `True`

`is` checks **object identity**, not value.

`x` is literally the same object as itself:

```python
x is x
```

→ **True**

###  Key Takeaway

```text
==  → checks values
is  → checks identity (same object)
```

For `NaN`:

```python
x == x    # False
x is x    # True
```

> **`NaN` is the classic example where `x == x` and `x is x` do not agree.**
