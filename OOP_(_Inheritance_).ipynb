# **Inheritance**

### 1. Basic Inheritance

A subclass inherits attributes and methods from its parent class.

```python
class Animal:
    def speak(self):
        print("Animal sound")


class Dog(Animal):
    pass


d = Dog()
d.speak()
```

```text
Dog → inherits from Animal
```

---

### 2. `super()`

`super()` lets a subclass reuse the parent's methods.

```python
class Animal:
    def __init__(self, name):
        self.name = name


class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name)
        self.breed = breed
```

Here:

```python
super().__init__(name)
```

runs the parent's `__init__()` instead of duplicating its code.

---

### 3. Method Overriding

A subclass can define a method with the same name as the parent.

```python
class Animal:
    def speak(self):
        print("Animal sound")


class Dog(Animal):
    def speak(self):
        print("Woof!")


d = Dog()
d.speak()
```

Output:

```text
Woof!
```

The subclass's method takes priority.

---

### 4. Extending a Parent Method

You don't always have to completely replace the parent's behavior.

```python
class Animal:
    def speak(self):
        print("Animal sound")


class Dog(Animal):
    def speak(self):
        super().speak()
        print("Woof!")
```

Output:

```text
Animal sound
Woof!
```

`super().speak()` runs the parent's version, then the subclass adds extra behavior.

---

### 5. `isinstance()` vs `type()`

```python
class Animal:
    pass

class Dog(Animal):
    pass

d = Dog()
```

```python
isinstance(d, Dog)      # True
isinstance(d, Animal)   # True
type(d) == Dog          # True
type(d) == Animal       # False
```

### Key Difference

```text
isinstance() → considers inheritance
type()       → checks the exact type
```

---

### 6. Multi-Level Inheritance

Inheritance can form a chain:

```python
class A:
    def show(self):
        print("A")


class B(A):
    pass


class C(B):
    pass
```

`C` inherits from `B`, and `B` inherits from `A`.

Therefore:

```python
c = C()
c.show()
```

works because Python searches upward through the inheritance chain.

```text
C → B → A
```

---

### 7. MRO — Method Resolution Order

MRO is the order Python follows when searching for a method or attribute.

```python
class A:
    pass

class B:
    pass

class C(A, B):
    pass

print(C.__mro__)
```

Python searches according to the MRO.

You can also use:

```python
C.mro()
```

MRO becomes especially important with **multiple inheritance**.

---

### 8. Overriding `__init__()` Without `super()`

This is a common mistake:

```python
class Animal:
    def __init__(self, name):
        self.name = name


class Dog(Animal):
    def __init__(self, breed):
        self.breed = breed
```

Now:

```python
d = Dog("Labrador")
```

`Animal.__init__()` never runs, so:

```python
d.name
```

does not exist.

### Correct approach:

```python
class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name)
        self.breed = breed
```

###  Key Takeaway

> **Inheritance lets a child reuse parent behavior. `super()` lets you explicitly use that parent behavior, while overriding lets the child customize it.**

Remember these three ideas:

```text
Inheritance → reuse
super()     → access parent behavior
Overriding  → customize behavior
```

````python
class Animal:
    def speak(self):
        return "Some sound"

class Dog(Animal):
    pass

d = Dog()
print(d.speak())

# Some sound
````

````python
class Animal:
    def speak(self):
        return "Some sound"

class Dog(Animal):
    def speak(self):
        return "Woof"

d = Dog()
print(d.speak())

# Woof
````

````python
class Animal:
    def __init__(self, name):
        self.name = name

class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name)
        self.breed = breed

d = Dog("Rex", "Labrador")
print(d.name, d.breed)

# Rex Labrador
````

````python
class Animal:
    def __init__(self, name):
        self.name = name

class Dog(Animal):
    pass

d = Dog("Buddy")
print(isinstance(d, Dog))
print(isinstance(d, Animal))

# True
# False
````

````python
class A:
    def greet(self):
        return "Hello from A"

class B(A):
    def greet(self):
        return super().greet() + " and B"

class C(B):
    def greet(self):
        return super().greet() + " and C"

c = C()
print(c.greet())

# Hello from A and B and C
````

````python
class Vehicle:
    wheels = 4

class Motorcycle(Vehicle):
    wheels = 2

class Car(Vehicle):
    pass

print(Motorcycle().wheels)
print(Car().wheels)
print(Vehicle().wheels)

# 2
# 4
# 4
````

````python
class Shape:
    def area(self):
        raise NotImplementedError("Subclass must implement this")

class Rectangle(Shape):
    def __init__(self, w, h):
        self.w = w
        self.h = h
    def area(self):
        return self.w * self.h

class Square(Rectangle):
    def __init__(self, side):
        super().__init__(side, side)

s = Square(4)
print(s.area())
print(isinstance(s, Shape))

# 16
# True
````

````python
class Bird:
    def __init__(self, name):
        self.name = name
    def fly(self):
        return f"{self.name} flies"

class Penguin(Bird):
    def fly(self):
        return f"{self.name} cannot fly"

birds = [Bird("Sparrow"), Penguin("Pingu")]
for b in birds:
    print(b.fly())

# 'Sparrow flies' 'Pingu cannot fly'
````

````python
class Base:
    def __init__(self):
        self.value = self.get_value()
    def get_value(self):
        return 10

class Derived(Base):
    def get_value(self):
        return 20

d = Derived()
print(d.value)

# 20
````

````python
class A:
    x = 1
    def __init__(self):
        A.x += 1

a1 = A()
a2 = A()
a3 = A()

print(A.x)
print(a1.x, a2.x, a3.x)

# 4
# 2 3 4
````

# Class Attribute vs Instance Attribute

```python
class A:
    x = 1

    def __init__(self):
        A.x += 1


a1 = A()
a2 = A()
a3 = A()

print(A.x)
print(a1.x, a2.x, a3.x)
```

### Output

```text
4
4 4 4
```

### Why?

Initially:

```text
A.x = 1
```

Each object creation runs:

```python
A.x += 1
```

So:

```text
a1 → x becomes 2
a2 → x becomes 3
a3 → x becomes 4
```

Therefore:

```python
print(A.x)
```

outputs:

```text
4
```

But **no instance ever gets its own `x`** because there is no:

```python
self.x = ...
```

So:

```python
a1.x
a2.x
a3.x
```

all fall back to the **same class attribute** `A.x`.

At print time:

```text
A.x = 4
```

Therefore:

```text
a1.x = 4
a2.x = 4
a3.x = 4
```

###  Key Takeaway

> **Class attributes are shared. If an instance doesn't have its own attribute, Python looks up the class attribute and gets its current value.**

```text
A.x
 ↑
 ├── a1.x
 ├── a2.x
 └── a3.x
```

All three point to the same class-level value unless `self.x` is created.

````python
class Logger:
    def __init__(self):
        print(f"Logger init, class = {self.__class__.__name__}")

class FileLogger(Logger):
    def __init__(self):
        super().__init__()
        print("FileLogger init")

f = FileLogger()

# Logger init, class = Logger
# FileLogger init
````

# `self.__class__` — Actual Runtime Type

```python
class Logger:
    def __init__(self):
        print(f"Logger init, class = {self.__class__.__name__}")


class FileLogger(Logger):
    def __init__(self):
        super().__init__()
        print("FileLogger init")


f = FileLogger()
```

### Output

```text
Logger init, class = FileLogger
FileLogger init
```

### Why?

When:

```python
f = FileLogger()
```

is created, `self` is actually a **`FileLogger` object**.

Then:

```python
super().__init__()
```

calls `Logger.__init__()`.

Even though the code is running inside `Logger`:

```python
self.__class__.__name__
```

looks at the **actual runtime type of `self`**.

Therefore:

```python
self.__class__        # FileLogger
self.__class__.__name__  # "FileLogger"
```

### Key Takeaway

> **`self.__class__` tells you the actual class of the object, not the class whose method is currently executing.**

```text
FileLogger object
       ↓
Logger.__init__()
       ↓
self → FileLogger instance
       ↓
self.__class__.__name__
       ↓
"FileLogger"
```

````python
class A:
    def method(self):
        return "A"

class B(A):
    def method(self):
        return "B"

class C(A):
    def method(self):
        return "C"

class D(B, C):
    pass

d = D()
print(d.method())
print(D.__mro__)

# B
# C
````

# Multiple Inheritance & MRO

```python
class A:
    def method(self):
        print("A")


class B(A):
    def method(self):
        print("B")


class C(A):
    def method(self):
        print("C")


class D(B, C):
    pass


d = D()

d.method()
print(D.__mro__)
```

### Output

```text
B
(<class 'D'>, <class 'B'>, <class 'C'>, <class 'A'>, <class 'object'>)
```

### Why does `d.method()` print `B`?

`D` inherits from:

```python
class D(B, C):
```

Python searches the classes according to the **MRO**.

```text
D → B → C → A → object
```

`B` comes before `C`, and `B` already has its own `method()`.

Therefore:

```python
d.method()
```

uses:

```python
B.method()
```

and prints:

```text
B
```

### What does `D.__mro__` show?

`D.__mro__` returns a **tuple containing the complete method resolution order**:

```text
D → B → C → A → object
```

So:

```python
print(D.__mro__)
```

produces:

```text
(<class 'D'>, <class 'B'>, <class 'C'>, <class 'A'>, <class 'object'>)
```

### Key Takeaway

> **MRO is the order Python follows when searching for methods and attributes.**

```text
D
↓
B  ← method found here
↓
C
↓
A
↓
object
```

`A` appears only **once**, even though both `B` and `C` inherit from it. Python uses **C3 linearization** to create a consistent MRO and avoid ambiguity in diamond inheritance.

````python
class Base:
    def __init__(self):
        print("Base init")

class Child(Base):
    def __init__(self):
        print("Child init")

c = Child()

# Child init
````

````python
class Account:
    def __init__(self, balance=0):
        self.balance = balance
        self.history = []

class SavingsAccount(Account):
    def __init__(self, balance=0, rate=0.05):
        self.rate = rate
        # forgot to call super().__init__()

s = SavingsAccount(100)
print(s.rate)
print(s.balance)

````
